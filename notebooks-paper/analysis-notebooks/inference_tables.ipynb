```python
from src.paths import REPO_ROOT
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

import os

from PIL import Image
from pathlib import Path
from typing import List, Tuple, Dict

```

```python
def print_table_from_csv(csv_path: str):
    df = pd.read_csv(csv_path)

    models = df['model'].unique().tolist()
    testset = df['testset'].unique().tolist()

    # Aggregated results on different models accross test sets
    for model in models:
        df_model = df[df['model'] == model]
        print(f"Model: {model}")
        for metric in ['psnr', 'ssim', 'gmsd']:
            mean_metric = df_model[metric].mean()
            std_metric = df_model[metric].std()

            std_err = stats.sem(df_model[metric])
            ci_95 = std_err * stats.t.ppf(0.975, len(df_model[metric]) - 1)
            print(f"  {metric.upper()}: {mean_metric:.4f} ± {ci_95:.4f}, std: {std_metric:.4f}")
```

```python
# Case 4 uspforp with different noise models
df_nm0 = pd.read_csv(f'{REPO_ROOT}/notebooks-paper/data/case4_uspforp_nm0.csv')
df_nm0['noise_model'] = 0

df_nm1 = pd.read_csv(f'{REPO_ROOT}/notebooks-paper/data/case4_uspforp_nm1.csv')
df_nm1['noise_model'] = 1

df_nm2 = pd.read_csv(f'{REPO_ROOT}/notebooks-paper/data/case4_uspforp_nm2.csv')
df_nm2['noise_model'] = 2

df_nm3 = pd.read_csv(f'{REPO_ROOT}/notebooks-paper/data/case4_uspforp_nm3.csv')
df_nm3['noise_model'] = 3

df_nm4 = pd.read_csv(f'{REPO_ROOT}/notebooks-paper/data/case4_uspforp_nm4.csv')
df_nm4['noise_model'] = 4

df_case4_uspforp = pd.concat([df_nm0, df_nm1, df_nm2, df_nm3, df_nm4], ignore_index=True)

models_case4_uspforp = df_case4_uspforp['model'].unique().tolist()

# Aggregated results on different models accross test sets
for model in models_case4_uspforp:
    df_model = df_case4_uspforp[df_case4_uspforp['model'] == model]
    print(f"Model: {model}")
    for metric in ['psnr', 'ssim', 'lpips', 'gmsd']:
        mean_metric = df_model[metric].mean()
        std_metric = df_model[metric].std()

        std_err = stats.sem(df_model[metric])
        ci_95 = std_err * stats.t.ppf(0.975, len(df_model[metric]) - 1)
        print(f"  {metric.upper()}: {mean_metric:.4f} ± {ci_95:.4f}, std: {std_metric:.4f}")

```
