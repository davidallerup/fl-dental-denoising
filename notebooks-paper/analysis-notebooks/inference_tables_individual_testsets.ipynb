```python
from src.paths import REPO_ROOT
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

import os

from PIL import Image
from pathlib import Path
from typing import List, Tuple, Dict

```

```python
def print_table_from_csv_on_testset(csv_path: str, testset: int):
    df = pd.read_csv(csv_path)

    models = df['model'].unique().tolist()

    # Aggregated results on different models accross test sets
    for model in models:
        df_model = df[(df['model'] == model) & (df['testset'] == testset)]
        print(f"Model: {model}")
        for metric in ['psnr', 'ssim', 'gmsd']:
            mean_metric = df_model[metric].mean()
            std_metric = df_model[metric].std()

            std_err = stats.sem(df_model[metric])
            ci_95 = std_err * stats.t.ppf(0.975, len(df_model[metric]) - 1)
            print(f"  {metric.upper()}: {mean_metric:.4f} ± {ci_95:.4f}, std: {std_metric:.4f}")
```

```python
# Case 2
print_table_from_csv_on_testset(f'{REPO_ROOT}/notebooks-paper/data/case2.csv', testset=2)
```

```python
# Case 2 uspforp
print_table_from_csv_on_testset(f'{REPO_ROOT}/notebooks-paper/data/case2_uspforp.csv', testset=0)
```
